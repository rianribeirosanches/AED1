```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Carregar a base de dados pública do Titanic
titanic_df = sns.load_dataset('titanic')

# Exibir as primeiras linhas do DataFrame para verificar
print("Primeiras 5 linhas do DataFrame Titanic:")
print(titanic_df.head())

# Criar um gráfico de barras para visualizar a contagem de mortos e sobreviventes por gênero
plt.figure(figsize=(8, 6))
sns.countplot(data=titanic_df, x='sex', hue='survived', palette='viridis')
plt.title('Sobreviventes e Mortos por Gênero no Titanic')
plt.xlabel('Gênero')
plt.ylabel('Número de Pessoas')
plt.xticks(ticks=[0, 1], labels=['Masculino', 'Feminino'])
plt.legend(title='Sobreviveu', labels=['Não', 'Sim'])
plt.grid(axis='y', linestyle='--')
plt.show()
```
